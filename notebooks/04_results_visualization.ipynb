## 1. Setup

```python
import sys
import os
sys.path.insert(0, os.path.abspath('../src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

from utils import VisualizationHelper

VisualizationHelper.set_style()
%matplotlib inline

print("Libraries loaded successfully!")
```

## 2. Load Results

```python
# Load autocomplete results
try:
    autocomplete_results = pd.read_csv('../results/autocomplete_comparison.csv', index_col=0)
    print("Autocomplete Results:")
    print(autocomplete_results)
except FileNotFoundError:
    print("Autocomplete results not found. Please run notebook 02 first.")
    autocomplete_results = None
```

```python
# Load autocorrect results
try:
    autocorrect_results = pd.read_csv('../results/autocorrect_comparison.csv', index_col=0)
    print("\nAutocorrect Results:")
    print(autocorrect_results)
except FileNotFoundError:
    print("Autocorrect results not found. Please run notebook 03 first.")
    autocorrect_results = None
```

## 3. Comprehensive Performance Analysis

```python
if autocomplete_results is not None:
    # Analyze autocomplete algorithms
    print("\n=== AUTOCOMPLETE ANALYSIS ===")
    print("\nBest Algorithm by Metric:")
    
    for col in autocomplete_results.columns:
        if 'time' in col.lower():
            best_idx = autocomplete_results[col].idxmin()
            print(f"  {col}: {best_idx} ({autocomplete_results.loc[best_idx, col]:.6f})")
        else:
            best_idx = autocomplete_results[col].idxmax()
            print(f"  {col}: {best_idx} ({autocomplete_results.loc[best_idx, col]:.6f})")
```

```python
if autocorrect_results is not None:
    print("\n=== AUTOCORRECT ANALYSIS ===")
    print("\nBest Algorithm by Metric:")
    
    for col in autocorrect_results.columns:
        if 'time' in col.lower() or 'error' in col.lower():
            best_idx = autocorrect_results[col].idxmin()
            print(f"  {col}: {best_idx} ({autocorrect_results.loc[best_idx, col]:.6f})")
        else:
            best_idx = autocorrect_results[col].idxmax()
            print(f"  {col}: {best_idx} ({autocorrect_results.loc[best_idx, col]:.6f})")
```

## 4. Create Summary Dashboard

```python
# Create comprehensive comparison visualizations
if autocomplete_results is not None and autocorrect_results is not None:
    fig = plt.figure(figsize=(16, 10))
    gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)
    
    # Row 1: Autocomplete metrics
    # MRR
    ax1 = fig.add_subplot(gs[0, 0])
    if 'mean_reciprocal_rank' in autocomplete_results.columns:
        autocomplete_results['mean_reciprocal_rank'].plot(kind='bar', ax=ax1, color='steelblue')
        ax1.set_title('Autocomplete: Mean Reciprocal Rank')
        ax1.set_ylabel('MRR')
    
    # Precision@5
    ax2 = fig.add_subplot(gs[0, 1])
    if 'precision_at_5' in autocomplete_results.columns:
        autocomplete_results['precision_at_5'].plot(kind='bar', ax=ax2, color='darkgreen')
        ax2.set_title('Autocomplete: Precision@5')
        ax2.set_ylabel('Precision')
    
    # Query Time
    ax3 = fig.add_subplot(gs[0, 2])
    if 'avg_query_time' in autocomplete_results.columns:
        (autocomplete_results['avg_query_time']*1000).plot(kind='bar', ax=ax3, color='coral')
        ax3.set_title('Autocomplete: Avg Query Time')
        ax3.set_ylabel('Time (ms)')
    
    # Row 2: Autocorrect metrics
    # Accuracy
    ax4 = fig.add_subplot(gs[1, 0])
    if 'accuracy' in autocorrect_results.columns:
        autocorrect_results['accuracy'].plot(kind='bar', ax=ax4, color='steelblue')
        ax4.set_title('Autocorrect: Accuracy')
        ax4.set_ylabel('Accuracy')
        ax4.set_ylim(0, 1)
    
    # WER
    ax5 = fig.add_subplot(gs[1, 1])
    if 'avg_wer' in autocorrect_results.columns:
        autocorrect_results['avg_wer'].plot(kind='bar', ax=ax5, color='darkgreen')
        ax5.set_title('Autocorrect: Word Error Rate')
        ax5.set_ylabel('WER')
    
    # CER
    ax6 = fig.add_subplot(gs[1, 2])
    if 'avg_cer' in autocorrect_results.columns:
        autocorrect_results['avg_cer'].plot(kind='bar', ax=ax6, color='coral')
        ax6.set_title('Autocorrect: Character Error Rate')
        ax6.set_ylabel('CER')
    
    # Row 3: Combined efficiency analysis
    ax7 = fig.add_subplot(gs[2, :])
    
    # Training time comparison
    training_times_ac = autocomplete_results.get('training_time', pd.Series([0]*len(autocomplete_results)))
    training_times_acr = autocorrect_results.get('training_time', pd.Series([0]*len(autocorrect_results)))
    
    if training_times_ac.sum() > 0 or training_times_acr.sum() > 0:
        x = np.arange(len(autocomplete_results))
        width = 0.35
        ax7.bar(x - width/2, training_times_ac, width, label='Autocomplete', alpha=0.8)
        ax7.bar(x + width/2, training_times_acr[:len(autocomplete_results)], width, label='Autocorrect', alpha=0.8)
        ax7.set_xticks(x)
        ax7.set_xticklabels(autocomplete_results.index)
        ax7.set_title('Training Time Comparison')
        ax7.set_ylabel('Time (seconds)')
        ax7.legend()
    
    plt.suptitle('Autocomplete and Autocorrect Performance Dashboard', fontsize=16, y=1.00)
    plt.savefig('../results/performance_dashboard.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("✓ Dashboard saved to '../results/performance_dashboard.png'")
```
